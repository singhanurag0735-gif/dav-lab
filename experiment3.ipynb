```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# read the dataset
df = pd.read_csv("student_performance.csv")

# replace missing marks with average marks
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df.head())

# Histogram of marks
plt.hist(df["Marks"], bins=5, edgecolor="black")
plt.title("Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# Box plot of marks
sns.boxplot(y=df["Marks"])
plt.title("Marks Box Plot")
plt.show()

# Study hours and marks
sns.scatterplot(x=df["StudyHours"], y=df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()

# Pair plot
sns.pairplot(df[["StudyHours", "Attendance", "Marks"]])
plt.show()

# Correlation heatmap
corr = df[["StudyHours", "Attendance", "Marks"]].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
