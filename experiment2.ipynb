```python
import pandas as pd

# load the student data
df = pd.read_csv("students.csv")

print("Student Data")
print(df)

# basic information about the data
print("\nData Information")
print(df.info())

# descriptive statistics
print("\nDescriptive Statistics")
print(df.describe())

# checking missing values
print("\nMissing Values")
print(df.isnull().sum())

# checking duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

print("\nDuplicate Rows")
print(df[df.duplicated()])

# some basic calculations
print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

# average marks in each branch
print("\nAverage Marks by Branch")
print(df.groupby("Branch")["Marks"].mean())

# fee details
fee = pd.DataFrame({
    "Branch": ["CSE", "ISE", "CSD", "AIML"],
    "Annual Fee": [120000, 115000, 110000, 125000]
})

# merging student data with fee details
merged = pd.merge(df, fee, on="Branch", how="left")

print("\nStudent Data with Fee Details")
print(merged)

# pivot table
pivot = pd.pivot_table(
    df,
    values="Marks",
    index="Branch",
    columns="Result",
    aggfunc="count"
)

print("\nResult Count by Branch")
print(pivot)
```
